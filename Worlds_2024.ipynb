```python
import pandas as pd
from IPython.display import display, Markdown
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv('Worlds_MainEvent_2024.csv',sep=';')
with pd.option_context('display.max_columns', None):
    display(df.head(15))
```

```python
# Tạo DataFrame mới với các trận đấu duy nhất
match_details = df[['Date', 'Round', 'Format', 'No Game', 'Team', 'Opponent Team', 'Outcome', 'Side']].drop_duplicates().reset_index(drop=True)
match_details.tail()
# hiển thị match_details với outcome là win
match_details_win = match_details[match_details['Outcome']=='Win'].reset_index(drop=True)

games_results = pd.DataFrame({
    'Date': match_details_win['Date'],
    'Round': match_details_win['Round'],
    'Format': match_details_win['Format'],
    'No Game': match_details_win['No Game'],
    'Blue Team': match_details_win['Team'],
    'Red Team': match_details_win['Opponent Team'],
    'Side Win': match_details_win['Side']
})

# Thêm cột 'Team Win' dựa trên 'Side Win' (nếu Side Win == 'Blue' -> Team Win = Blue Team; nếu 'Red' -> Red Team)
def _resolve_team_win(row):
    side = row.get('Side Win')
    if pd.isna(side):
        return None
    side_str = str(side).strip().lower()
    if side_str == 'blue':
        return row.get('Blue Team')
    if side_str == 'red':
        return row.get('Red Team')
    return None

games_results['Team Win'] = games_results.apply(_resolve_team_win, axis=1)

display(Markdown("### Games Results - Win Matches"))
with pd.option_context('display.max_rows', None, 'display.max_columns', None):
    display(games_results)
```

```python
# Sửa lại bảng Games Results: xóa cột team Win cũ đi, Blue Team thành team Win, Red thành Team Lose, Side Win giữ nguyên
games_results_final = games_results.drop(columns=['Team Win'])
games_results_final = games_results_final.rename(columns={
    'Blue Team': 'Team Win',
    'Red Team': 'Team Lose'
})

# sau đó add 2 cột mới Blue Team và Red Team với điều kiện: nếu Side Win là Blue thì Blue Team = Team Win, Red Team = Team Lose; ngược lại nếu Side Win là Red thì Red Team = Team Win, Blue Team = Team Lose
def _resolve_teams(row):
    side = row.get('Side Win')
    team_win = row.get('Team Win')
    team_lose = row.get('Team Lose')
    if pd.isna(side):
        return pd.Series([None, None])
    side_str = str(side).strip().lower()
    if side_str == 'blue':
        return pd.Series([team_win, team_lose])
    if side_str == 'red':
        return pd.Series([team_lose, team_win])
    return pd.Series([None, None])
games_results_final[['Blue Team', 'Red Team']] = games_results_final.apply(_resolve_teams, axis=1)

# Sắp xếp lại cột theo thứ tự sau: Date, Round, Format, No Game, Blue Team, Red Team, Side Win, Team Win
games_results_final = games_results_final[['Date', 'Round', 'Format', 'No Game', 'Blue Team', 'Red Team', 'Side Win', 'Team Win']]
display(Markdown("### Games Results - Final Sorted Columns"))
with pd.option_context('display.max_rows', None, 'display.max_columns', None):
    display(games_results_final)

#Sửa lại vào file csv MainEvent_GameResults.csv
games_results_final.to_csv('MainEvent_GameResults.csv', index=False)
```

```python
# Tạo DataFrame mới với các trận đấu duy nhất
match_details = df[['Date', 'Round', 'Format', 'No Game', 'Team', 'Opponent Team', 'Outcome', 'Side']].copy()

# Lấy thông tin game cuối cùng của mỗi trận đấu
last_games = df.groupby(['Date', 'Round', 'Team', 'Opponent Team']).agg({
    'No Game': 'max',
    'Format': 'first'
}).reset_index()

# Lấy kết quả của game cuối cùng
final_outcomes = pd.merge(
    last_games,
    df[['Date', 'Round', 'Team', 'Opponent Team', 'No Game', 'Outcome', 'Side']],
    how='left',
    on=['Date', 'Round', 'Team', 'Opponent Team', 'No Game']
 )

# Tạo DataFrame kết quả
results = pd.DataFrame()
for _, game in final_outcomes.iterrows():
    if game['Outcome'] == 'Win':
        team_win = game['Team']
        # Determine result string based on Format and last No Game
        if game['Format'] == 1:
            result_win = '1-0'
            result_loss = '0-1'
        elif game['Format'] == 3:
            if game['No Game'] == 2:
                result_win = '2-0'
                result_loss = '0-2'
            else:  # No Game == 3
                result_win = '2-1'
                result_loss = '1-2'
        else:  # Format == 5
            if game['No Game'] == 3:
                result_win = '3-0'
                result_loss = '0-3'
            elif game['No Game'] == 4:
                result_win = '3-1'
                result_loss = '1-3'
            else:  # No Game == 5
                result_win = '3-2'
                result_loss = '2-3'
                
        # Add winning team's record
        results = pd.concat([results, pd.DataFrame({
            'Date': [game['Date']],
            'Round': [game['Round']],
            'Format': [game['Format']],
            'No Game': [game['No Game']],
            'Team': [game['Team']],
            'Opponent Team': [game['Opponent Team']],
            'Team Win': [team_win],
            'Result': [result_win]
        })])
        
        # Add losing team's record
        results = pd.concat([results, pd.DataFrame({
            'Date': [game['Date']],
            'Round': [game['Round']],
            'Format': [game['Format']],
            'No Game': [game['No Game']],
            'Team': [game['Opponent Team']],
            'Opponent Team': [game['Team']],
            'Team Win': [team_win],
            'Result': [result_loss]
        })])

# Định nghĩa thứ tự của rounds
round_order = ['1', '2', '3', '4', '5', 'QUATERFINALS', 'SEMIFINALS', 'FINALS']
round_order_dict = {round_name: i for i, round_name in enumerate(round_order)}

# Reset index và sắp xếp kết quả theo Date, Round (theo thứ tự đã định nghĩa), và Team
results = results.drop_duplicates().reset_index(drop=True)
results = results.sort_values(
    by=['Date', 'Round', 'Team'],
    key=lambda x: x.map(round_order_dict) if x.name == 'Round' else x
 )

# Loại bỏ cột 'Side Win' nếu tồn tại (được yêu cầu)
results = results.drop(columns=['Side Win'], errors='ignore')

# Hiển thị kết quả
display(Markdown('## Match Results'))
with pd.option_context('display.max_rows', None, 'display.max_columns', None):
    display(results)

# Lưu kết quả vào file CSV
results.to_csv('MainEvent_MatchResults.csv', index=False)
```

```python
# Hàm tính tổng số trận đã chơi của một đội theo tên đội
def get_TotalGamesPlayedNumbyTeamName(team_name):
    team_games = df[df['Team'] == team_name]
    games_played = int(team_games['No Game'].count()/5)
    return games_played
    
get_TotalGamesPlayedNumbyTeamName('T1')        
```

```python
# Hàm tính tổng số trận thắng của một đội theo tên đội
def get_TotalGamesWonNumbyTeamName(team_name):
    team_wins = df[(df['Team'] == team_name) & (df['Outcome'] == 'Win')]
    games_won = int(team_wins['No Game'].count()/5)
    return games_won
get_TotalGamesWonNumbyTeamName('T1')
```

```python
# Hàm tính tổng số trận, số trận thắng và tỉ lệ thắng của một đội theo tên đội
def get_TeamWinrate(team_name):
    total_games = get_TotalGamesPlayedNumbyTeamName(team_name)
    games_won = get_TotalGamesWonNumbyTeamName(team_name)
    winrate = ((games_won / total_games) * 100)
    return winrate


team_winrate = pd.DataFrame({
    'Team': df['Team'].unique(),
    'Total Games': [get_TotalGamesPlayedNumbyTeamName(team) for team in df['Team'].unique()],
    'Games Won': [get_TotalGamesWonNumbyTeamName(team) for team in df['Team'].unique()],
    'Winrate (%)': [get_TeamWinrate(team) for team in df['Team'].unique()]
})
team_winrate = team_winrate.sort_values(by='Winrate (%)', ascending=False).reset_index(drop=True)
display(Markdown('## Team Winrate Summary'))
display(team_winrate)   
```

```python
def plot_ban_distribution(ban_data):
    # Count bans and ensure descending order (largest -> smallest)
    ban_counts = ban_data['Ban'].dropna().astype(str).apply(lambda s: [x.strip() for x in s.split(',') if x.strip()]).explode().value_counts().reset_index()
    ban_counts.columns = ['Champion', 'Ban Count']
    ban_counts = ban_counts.head(10).sort_values('Ban Count', ascending=True).reset_index(drop=True)

    fig = px.bar(
        ban_counts,
        y='Champion',
        x='Ban Count',
        title='Distribution of Banned Champions (descending)',
        labels={'Ban Count': 'Number of Bans', 'Champion': 'Champion'},
        color='Ban Count',
        color_continuous_scale='Viridis',
    )
    # Enforce category order so the largest banned champion appears at the top
    fig.update_layout(yaxis={'categoryorder':'array','categoryarray':ban_counts['Champion'].tolist()})
    fig.show()

plot_ban_distribution(ban)
```

```python
def top5_bans_by_team(team_name, topn=5):
    # Lọc dữ liệu cho team được chỉ định
    team_bans = df[df['Team'] == team_name]['Ban'].dropna()
    
    # Tách các tướng bị ban và đếm số lần xuất hiện
    ban_counts = team_bans.astype(str).apply(lambda s: [x.strip() for x in s.split(',') if x.strip()]).explode().value_counts()
    
    # Tạo DataFrame với top N tướng bị ban nhiều nhất
    top_bans = pd.DataFrame({
        'Champion': ban_counts.index[:topn],
        'Ban Count': ban_counts.values[:topn]
    })
    
    # Hiển thị kết quả
    display(Markdown(f'## Top {topn} Most Banned Champions by {team_name}'))
    display(top_bans)
    
    # Tạo biểu đồ
    fig = px.bar(
        top_bans,
        y='Champion',
        x='Ban Count',
        title=f'Top {topn} Most Banned Champions by {team_name}',
        labels={'Ban Count': 'Number of Bans', 'Champion': 'Champion'},
        color='Ban Count',
        color_continuous_scale='Viridis',
    )
    # Sắp xếp theo thứ tự giảm dần
    fig.update_layout(yaxis={'categoryorder':'array','categoryarray':top_bans['Champion'].tolist()[::-1]})
    fig.show()

# Hiển thị top 5 tướng bị ban nhiều nhất của T1
top5_bans_by_team('GEN')
```

```python
#Hiển thị bảng dữ liệu các player gồm trung bình các chỉ số : Player, Role, Team, Kills, Deaths, Assists, KDA ( = (Kills + Assists)/Deaths), CS, Golds , GPM, %Gold, VisionScore, %XP Share , DPM, DMG%, KP%, SoloKills (Tổng), Double kills(Tổng), Triple kills(Tổng), Quadra kills(Tổng), Penta kills(Tổng), Games Played, Wins, Win Rate (%)
player_stats = df[['Player', 'Role', 'Team', 'Kills', 'Deaths', 'Assists', 'CS', 'Golds', 'GPM', 'GOLD%', 'Vision Score', 'DPM', 'DMG%', 'KP%', 'Solo kills', 'Double kills', 'Triple kills', 'Quadra kills', 'Penta kills']].copy()
player_stats['KDA'] = (player_stats['Kills'] + player_stats['Assists']) / player_stats['Deaths'].replace(0, 1)  # Tránh chia cho 0

#tính trung bình các chỉ số cho mỗi player, trừng các chỉ số Solo kills, Double kills, Triple kills, Quadra kills, Penta kills, Games Played, Wins, Win Rate (%) và lấy 2 chữ số thập phân
player_stats = player_stats.groupby(['Player', 'Role', 'Team']).agg({
    'Kills': 'mean',
    'Deaths': 'mean',
    'Assists': 'mean',
    'KDA': 'mean',
    'CS': 'mean',
    'Golds': 'mean',
    'GPM': 'mean',
    'GOLD%': 'mean',
    'Vision Score': 'mean',
    'DPM': 'mean',
    'DMG%': 'mean',
    'KP%': 'mean',
    'Solo kills': 'sum',
    'Double kills': 'sum',
    'Triple kills': 'sum',
    'Quadra kills': 'sum',
    'Penta kills': 'sum'
}).round(2).reset_index()
# Thêm cột Games Played, Wins, Win Rate (%)
def get_player_games_played(player_name):
    return int(df[df['Player'] == player_name]['No Game'].count())
def get_player_wins(player_name):
    return int(df[(df['Player'] == player_name) & (df['Outcome'] == 'Win')]['No Game'].count()) 
player_stats['Games Played'] = player_stats['Player'].apply(get_player_games_played)
player_stats['Wins'] = player_stats['Player'].apply(get_player_wins)
player_stats['Win Rate (%)'] = (player_stats['Wins'] / player_stats['Games Played']).round(2)

display(Markdown('## Player Stats Summary'))
with pd.option_context('display.max_rows', None, 'display.max_columns', None):
    display(player_stats)

# Xuất bảng player_stats ra file CSV
player_stats.to_csv('MainEvent_PlayerStats.csv', index=False)
```

```python
    #TÍnh lại cột KDA = (Kills + Assists) / Deaths, nếu Deaths = 0 thì KDA = Kills + Assists sau đó sửa lại vào file Worlds_MainEvent_2024.csv
def recalculate_kda(row):
    kills = row['Kills']
    assists = row['Assists']
    deaths = row['Deaths']
    if deaths == 0:
        return kills + assists
    else:
        return (kills + assists) / deaths
df['KDA'] = df.apply(recalculate_kda, axis=1)
# Lưu lại vào file CSV
df.to_csv('Worlds_MainEvent_2024.csv', index=False)
```
